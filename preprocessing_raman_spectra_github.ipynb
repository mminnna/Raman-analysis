```python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.append('D:\\Raman analysis files')
from mylibs import raman_preprocessing_library as rpl 
from tqdm.notebook import tqdm
import h5py
```

# Download data

```python
# Load Raman data from the h5 file back as dictionaries and verify the contents

path = 'D:\\Raman analysis files\\Substudy II analysis\\Shared Raman data\\raman_dataset.h5'

numpy_arrays_RA = {}
numpy_arrays_control = {}

with h5py.File(path, "r") as f:
    wavenumbers = f["wavenumbers"][:]
    ra_group = f["RA"]
    control_group = f["Control"]

    for sample_id in ra_group:
        numpy_arrays_RA[sample_id] = ra_group[sample_id][:]

    for sample_id in control_group:
        numpy_arrays_control[sample_id] = control_group[sample_id][:] 
```

```python
print("\nControl Group Data:")
for sample_id, data in numpy_arrays_control.items():
    # Remove rows containing any NaN
    data = data[~np.isnan(data).any(axis=1)]
    numpy_arrays_control[sample_id] = data  # update the dict
    print(f"Sample {sample_id}: Data shape {data.shape}")

print("\nRA Group Data:")
for sample_id, data in numpy_arrays_RA.items():
    # Remove rows containing any NaN
    data = data[~np.isnan(data).any(axis=1)]
    numpy_arrays_RA[sample_id] = data  # update the dict
    print(f"Sample {sample_id}: Data shape {data.shape}")
```

# Preprocessing

Remove outliers and NaN values from the data

```python
# Check if there are any NaN values left in data and remove those rows
for sample_id, data in numpy_arrays_control.items():
    data = data[~np.isnan(data).any(axis=1)]
    numpy_arrays_control[sample_id] = data  # update the dict
    # print the updated shape
    print(f"Control Sample {sample_id}: Data shape {data.shape}")

for sample_id, data in numpy_arrays_RA.items():
    data = data[~np.isnan(data).any(axis=1)]
    numpy_arrays_RA[sample_id] = data  # update the dict
    # print the updated shape
    print(f"RA Sample {sample_id}: Data shape {data.shape}")

```

Remove outliers

```python
# Remove outliers from each sample
data_control_filtered, data_control_rejected = rpl.remove_outliers_pca(numpy_arrays_control, wavenumbers, n_components=2, confidence_threshold=0.995, w0=550, w1=1750)
data_RA_filtered, data_RA_rejected = rpl.remove_outliers_pca(numpy_arrays_RA, wavenumbers, n_components=2, confidence_threshold=0.995, w0=550, w1=1750)

#Print shapes
print("\nFiltered Control Group Data:")
for sample_id, data in data_control_filtered.items():
   print(f"Sample {sample_id}: Data shape {data.shape}")

print("\nFiltered RA Group Data:")
for sample_id, data in data_RA_filtered.items():
   print(f"Sample {sample_id}: Data shape {data.shape}")

# Print the number of rejected spectra for each sample
print("\nNumber of rejected spectra in Control group:")
for sample_id, data in data_control_rejected.items():
   print(f"Sample {sample_id}: Number of rejected spectra {data.shape[0]}")
print("\nNumber of rejected spectra in RA group:")
for sample_id, data in data_RA_rejected.items():
   print(f"Sample {sample_id}: Number of rejected spectra {data.shape[0]}")
```

Match the number of spectra across samples after outlier removal

```python
# Match the number of spectra across samples by selecting a random subset of spectra from each sample
data_control_selected = rpl.select_random_spectra(data_control_filtered, 450)
data_RA_selected = rpl.select_random_spectra(data_RA_filtered, 450)

# Print shapes for verification
print("\nSelected Control data:")
for sample_id, data in data_control_selected.items():
   print(f"Sample {sample_id}: Data shape {data.shape}")

print("\nSelected RA data:")
for sample_id, data in data_RA_selected.items():
   print(f"Sample {sample_id}: Data shape {data.shape}")
```

Truncate the data into biofingerprint region

```python

data_truncated_control, truncated_wavenumbers = rpl.truncate_spectra(data_control_selected, wavenumbers, low_limit=500, high_limit=1700)
data_truncated_RA, _= rpl.truncate_spectra(data_RA_selected, wavenumbers, low_limit=500, high_limit=1700)
```

Remove cosmic spikes

```python
# Remove cosmic spikes
csr = rpl.CosmicSpikeRemoval()
control_data_cosmic_spikes_removed = csr.filter_cosmic_rays(
    input_dict=data_truncated_control,
    width_threshold=4,
    prominence_threshold=1000
)

RA_data_cosmic_spikes_removed = csr.filter_cosmic_rays(
    input_dict=data_truncated_RA,
    width_threshold=4,
    prominence_threshold=1000
)
```

```python
# Smooth the spectral data
control_smoothed_dictionary = rpl.smooth_spectra(control_data_cosmic_spikes_removed, window_length=15, polyorder=3)
RA_smoothed_dictionary = rpl.smooth_spectra(RA_data_cosmic_spikes_removed, window_length=15, polyorder=3)
```

```python
# Baseline correction using AirPLS

bls = rpl.BaseLineCorrection(method='airpls', lambda_=1e5, porder=2, itermax=50)
control_baseline_corrected_dictionary = bls.correct_dict(control_smoothed_dictionary)
RA_baseline_corrected_dictionary = bls.correct_dict(RA_smoothed_dictionary)
```

```python
# Normalize the baseline corrected spectra and plot the mean spectra of both groups
norm = rpl.Normalization(normalization_type='vector')

control_normalized_dictionary = norm.normalize_dict(control_baseline_corrected_dictionary)
RA_normalized_dictionary = norm.normalize_dict(RA_baseline_corrected_dictionary)
```

```python
# Combine the spectra from control and RA groups into a single dictionary
data_dictionary = {
    **control_normalized_dictionary, # Include control samples
    **RA_normalized_dictionary       # Include RA samples
}

# Initialize a sample ID vector of the correct size
total_rows = sum(spectra.shape[0] for spectra in data_dictionary.values())
sample_id_vector= np.zeros(total_rows, dtype=int)

# Assign unique IDs to each sample
start_idx = 0  # Starting index for each sample's rows
for sample_id, spectra in data_dictionary.items():
    rows, _ = spectra.shape  # Get the number of rows for the current sample
    sample_id_vector[start_idx : start_idx + rows] = sample_id  # Assign IDs
    start_idx += rows  # Update starting index for the next sample

# Print the final ID vector
print(sample_id_vector)
 


all_data = []
for sample_id, spectra in data_dictionary.items():
    sample_ids = np.full((spectra.shape[0], 1), int(sample_id))
    all_data.append(np.hstack((sample_ids, spectra)))

data_matrix = np.vstack(all_data)
# Remove the first column (sample IDs) from the data matrix to create a new matrix containing only the spectral data
spectral_data_matrix = data_matrix[:, 1:]
```

```python
#Initialize array for class labels
YY = np.zeros(data_matrix.shape[0])

sample_ids_deep_learning = data_matrix[:, 0]  # Extract sample IDs from the first column  

# Remove the first column (sample IDs) from the data matrix for training
XX = spectral_data_matrix  # Exclude the first column (sample IDs)

# Initialize array for class labels
YY = np.zeros(data_matrix.shape[0])

# Define the number of rows based on control samples in dictionary
N_control = sum(spectra.shape[0] for spectra in control_normalized_dictionary.values())
# Define the number of rows based on RA samples in dictionary
N_RA = sum(spectra.shape[0] for spectra in RA_normalized_dictionary.values())

# Define the indices for control and RA classes
Control_indices = np.arange(N_control)
RA_indices = np.arange(N_control, N_control + N_RA)

# Assign labels: 0 = control, 1 = RA
YY[Control_indices] = 0
YY[RA_indices] = 1

# Convert YY to dtype int
YY = YY.astype(int)
print("labels array:", np.shape(YY))

print("shape of spectra array:", np.shape(XX))
print("YY:", YY)
```
